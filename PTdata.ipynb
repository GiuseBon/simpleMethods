```python
import numpy
from scipy import stats
from sklearn import linear_model
import csv
import pandas as pd
import matplotlib.pyplot as plt
import math
```

```python
df = pd.read_csv("menu-breakdown-may1-27-2021.csv",delimiter = ",",index_col = False)
```

```python
df.dropna(subset = ["Item Name"],inplace = True)
df = df.set_index("Item Name",drop=True, append=False, inplace=False, verify_integrity=False)
```

```python
plt.figure()
plt.scatter(df["Avg Price"],100*df["Quantity"]/sum(df["Quantity"]))
plt.show()
```

```python
plot = df.plot.pie(y = "Net Sales", figsize = (10,10),legend = False)
```

```python
r = [0]*100
for index, row in df.iterrows():
    x = math.ceil(row["Avg Price"])
    y = row["Quantity"]
    
    r[x] = r[x] + y
```

```python
s = sum(r)
for i in range(len(r)):
    r[i] = 100*r[i]/s
```

```python
plt.figure()
plt.plot(r)
plt.show()
z = list(range(100))
```

```python
reg = stats.linregress(z,r)
```
